```python
import re
import nltk
import pandas as pd
import gensim

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.tokenize import word_tokenize

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
```

### Task 1: Cosine Similarity

```python
docs = ['This is the first document.',
        'This document is the second document.',
        'And this is the third one.',
        'Is this the first document?']

vector_matrix = TfidfVectorizer().fit_transform(docs)
pd.DataFrame(cosine_similarity(vector_matrix), index=docs, columns=docs).round(3)
```

Sentences 1 and 4 are identical (1.0) because they contain the same words; sentence 3 is the least similar.

### Task 2: TF-IDF

```python
sentences = ['data science is one of the most important fields of science',
             'this is one of the best data science courses',
             'data scientists analyze data']

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(sentences)
df_tfidf = pd.DataFrame(result.toarray(), columns=tfidf.get_feature_names_out()).round(3)
df_tfidf
```

```python
for i, row in df_tfidf.iterrows():
    print(f"Sentence {i + 1}:", row.sort_values(ascending=False).head(3).index.tolist())
```

### Task 3

```python
df = pd.read_csv('../dataset/simpsons_script_lines.csv')
df = df.dropna(subset=['spoken_words'])

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

tokens = [word_tokenize(clean_text(line)) for line in df['spoken_words']]

Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1, seed=42)
```

### Task 4

```python
print(Skip_gram_model.wv.most_similar('homer'))
print(Skip_gram_model.wv.most_similar('marge'))
print(Skip_gram_model.wv.most_similar('bart'))
```
