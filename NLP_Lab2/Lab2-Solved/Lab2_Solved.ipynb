```python
import os
import re
import nltk
import pandas as pd
import kagglehub

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
```

## Task 1

```python
path = kagglehub.dataset_download("seriousran/appletwittersentimenttexts")
df = pd.read_csv(os.path.join(path, "apple-twitter-sentiment-texts.csv"))

hashtags = [tag for text in df['text'] for tag in re.findall(r"#\w+", text.lower())]

print("Number of hashtags:", len(hashtags))
print(pd.Series(hashtags).value_counts().head(10))
```

## Task 2

```python
text = "This year is 2021"

pattern = re.compile(r"\d+")
print(type(pattern))

text = pattern.sub("2022", text)
print(text)
```

`re.compile()` compiles a regex string into a reusable `Pattern` object. It is useful because the
pattern is compiled once and can be reused many times, which is faster and cleaner.

## Task 3

```python
text = "a 11 b 2 3 c 4"
print(re.split(r"\d+", text))
```

`re.split()` splits a string at every match of the pattern and returns the remaining parts as a list.

## Task 4

```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = "I'm enjoying the NLP course!"

for token in nlp(text):
    print(token.text)
```

`spacy.load()` loads a trained language pipeline (tokenizer, tagger, parser, NER, ...) and returns an
`nlp` object that processes text.

```python
print(nltk.word_tokenize(text))
```

For this sentence both tokenizers give the same tokens. spaCy returns `Token` objects with extra
attributes (lemma, POS, ...), while NLTK returns plain strings.
